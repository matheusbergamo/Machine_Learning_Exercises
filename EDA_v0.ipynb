```python
f, (ax_box, ax_hist) = plt.subplots(            # f stands for figure, next are the axes (subplots) variables
    nrows=2,                                    # number of rows
    sharex=True,                                # they share the same x axies
    gridspec_kw={"height_ratios": (0.2, 1)}     # sets the relative heights
)

sns.boxplot(data=df, x="rating", ax=ax_box)
ax_box.axvline(mean, color='r', linestyle='--')
ax_box.axvline(median, color='g', linestyle='-')
ax_box.axvline(mode, color='b', linestyle='-')

sns.histplot(data=df, x="rating", ax=ax_hist, kde=True) # Kernel Density Estimate -> estimate of probability density 
ax_hist.axvline(mean, color='r', linestyle='--', label="Média")
ax_hist.axvline(median, color='g', linestyle='-', label="Mediana")
ax_hist.axvline(mode, color='b', linestyle='-', label="Moda")

plt.legend()
ax_box.set(xlabel='')
```

```python
def box_and_hist(df): # take a df column as an argument
    mean = df['rating'].mean()
    median = df['rating'].median()
    mode = df['rating'].mode().values[0]

    f, (ax_box, ax_hist) = plt.subplots(            # f stands for figure, next are the axes (subplots) variables
        nrows=2,                                    # number of rows
        sharex=True,                                # they share the same x axies
        gridspec_kw={"height_ratios": (0.2, 1)}     # sets the relative heights
    )

    sns.boxplot(data=df, x="rating", ax=ax_box)
    ax_box.axvline(mean, color='r', linestyle='--')
    ax_box.axvline(median, color='g', linestyle='-')
    ax_box.axvline(mode, color='b', linestyle='-')

    sns.histplot(data=df, x="rating", ax=ax_hist, kde=False) # Kernel Density Estimate -> estimate of probability density 
    ax_hist.axvline(mean, color='r', linestyle='--', label="Média")
    ax_hist.axvline(median, color='g', linestyle='-', label="Mediana")
    ax_hist.axvline(mode, color='b', linestyle='-', label="Moda")

    plt.legend()
    ax_box.set(xlabel='')
```

```python
box_and_hist(df)
```
